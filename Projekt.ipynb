Damit keine Datensätze verloren gehen, werden die leeren Felder nicht gelöscht, sondern durch den Eintrag „No Data“ ersetzt.

Eine Ersetzung durch Mittelwerte ist hier nicht sinnvoll, da es sich bei den meisten dieser Spalten um Merkmale handelt, für die sich kein sinnvoller Durchschnitt berechnen lässt. Außerdem könnten dadurch falsche Werte entstehen, die nicht der tatsächlichen Situation entsprechen. Beispielsweise wäre es nicht sinnvoll, eine fehlende Uhrzeit oder Geschwindigkeitsbegrenzung durch einen Durchschnittswert zu ersetzen, da an einer bestimmten Stelle bereits eine feste Geschwindigkeitsbegrenzung gilt.
```python
#Ersetzung fehlender Werte durch "No Data"
df = df.fillna({"Road_Surface_Conditions": "No Data", "Road_Type": "No Data", "Time": "No Data", "Weather_Conditions": "No Data"})
```

Für weitere Analyse erweist es sich als sinnvoll eine weitere Spalte zu der Tabelle hinzuzufügen. Bei dieser Spalte handelt es sich um die Spalte "Season" und deren Werte sind die vier Jahreszeiten. Dabei richtet sich die Einteilung der Monate an die meteorologische Jahreszeiten an. Außerdem werden die einzelnen Tage in "Weekday" und "Weekend" unterteilt, damit man im weiter Verlauf untersuchen kann, ob eine der Kategorien in Bezug auf Unfälle bevorzugt wird.

```python
#Zuweisung der Monate den meteorologischen Jahreszeiten
jahreszeiten_zuweisung = {
    "Dec": "Winter", "Jan": "Winter", "Feb": "Winter",
    "Mar": "Spring", "Apr": "Spring", "May": "Spring",
    "Jun": "Summer", "Jul": "Summer", "Aug": "Summer",
    "Sep": "Autumn", "Oct": "Autumn", "Nov": "Autumn"}

df["Season"] = df["Month"].map(jahreszeiten_zuweisung)

#Zuweisung der Tage
tage_zuweisung = {"Monday": "Weekday", "Tuesday": "Weekday", "Wednesday": "Weekday", "Thursday": "Weekday", "Friday": "Weekday", "Saturday": "Weekend", "Sunday": "Weekend"}

df["Week_Part"] = df["Day_of_Week"].map(tage_zuweisung)
```

Als letzten Schritt der Datenbereinigung werden die Uhrzeiten angepasst. Dabei werden die Minutenangaben entfernt, sodass nur noch die jeweilige Stunde erhalten bleibt. Dies ist für die weitere Analyse sinnvoller, da vor allem untersucht werden soll, zu welchen Uhrzeiten die meisten Unfälle passieren. Es ist dabei weniger wichtig, ob ein Unfall beispielsweise um 14:10 Uhr oder um 14:45 Uhr passiert ist. Beide Unfälle werden somit der gleichen Stunde zugeordnet. Dadurch lassen sich die Unfallzahlen besser nach Uhrzeiten vergleichen.  Ebenfalls wird noch nach Unterteilt, ob der Unfall Morgens, Mittags, Abends oder Nachts stattgefunden hat. Dazu wählen wir die folgende Einteilung:  Morgens 06:00 - 11:59 / Mittags 12.00 - 17:59 / Abends 18:00 - 21:59 / Nachts 22:00 - 05:59.

```python
#Funktion die eine gegebene Uhrzeit bei ":" teilt und die vorderste Zahl zurückgibt
def berechne_stunde(uhrzeit):
    #[0] am Ende wichtig, da ansonsten eine Liste im Format [XX, YY], bzw [X, YY] zurückgegeben wird
    return uhrzeit.split(":")[0]

#Ersetze alle alten Werte durch neue Werte, wobei auf jeden Eintrag die Funktion berechne_stunde angewandt wird
df["Time"] = df["Time"].apply(berechne_stunde)

#Funktion für die Zuweisung der Tageszeiten
def tageszeit(uhrzeit):
    if uhrzeit == "No Data":
        return "No Data"
    elif 6 <= int(uhrzeit) < 12:
        return "Morning"
    elif 12 <= int(uhrzeit) < 18:
        return "Afternoon"
    elif 18 <= int(uhrzeit) < 22:
        return "Evening"
    else:
        return "Night"

df["Day_Period"] = df["Time"].map(tageszeit)
```

```python
#Funktion welche die verschiedenen Cramer V's berechnet
def cramers_v_matrix(df, cols):
    #Erstellung einer Matrix die mit 1-en gefüll wird
    matrix = pd.DataFrame(np.ones((len(cols), len(cols))), index = cols, columns = cols)
    #Iterationen über die Spalten
    for col1 in cols:
        for col2 in cols:
            if col1 != col2:
                ct = pd.crosstab(df[col1], df[col2])
                matrix.loc[col1, col2] = association(ct, method="cramer")
    return matrix

nom_cols = ['Accident_Severity', 'Road_Type', 'Weather_Conditions', 'Road_Surface_Conditions', 'Urban_or_Rural_Area', 'Month', 'Day_of_Week', 'Time', 'Carriageway_Hazards', 'Junction_Control', 'Light_Conditions', 'Vehicle_Type', 'Junction_Detail'] 
cramer_df = cramers_v_matrix(df, nom_cols)
cramer_df
```
