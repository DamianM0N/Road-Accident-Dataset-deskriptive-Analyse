Es gibt ein einziges Duplikat, dieses wird aus dem Datensatz gelöscht.

```python
#Löschung der Zeile mit dem Duplikat
df = df.drop_duplicates()
```

Ein bereits angesprochenes Problem ist, dass die Wochentage nicht zum jeweiligen Datum passen. Durch Umwandlung der Spalte in ein Pandas Datum-Format wird der richtige Wochentag zum jeweiligen Datum herausgefunden und dann ersetzt.

```python
"""
   Umwandlung der Spalten Einträge in ein Pandas Datum-Format
   dayfirst = False: Die erste Zahl im Datum entspricht nicht dem Tag, sondern dem Monat
   errors = cource: Wandelt fehlerhafte Einträge zu NaT (Not a Time)
"""
df["Accident Date"] = pd.to_datetime(df["Accident Date"], dayfirst = False, errors = "coerce")

#Ersetzung der falschen Wochentage
df["Day_of_Week"] = df["Accident Date"].dt.day_name()

#Ausgabe ob Werte fehlen
print(f"{df["Day_of_Week"].isna().sum()} sind ungültig.")
```

Mit der obigen Befehlsausführung wurden alle Wochentage mit den passenden Wochentag zu dem jeweiligen Datum ersetzt. Zum Schluss fand noch eine Überprüfung statt, ob es fehlende Werte gibt, die jedoch zeigt, dass alle Zeilen mit gültigen Wochentagen befüllt sind.

Als nächstes werden Spalten gelöscht. Dabei werden die folgenden Spalten entfernt:<br>
- "Accident_Index"
- "Accident Date"
- "Year"
- "Latitude"
- "Longitude"

Keine der Spalten ist für die weitere Datenanalyse von Bedeutung, da es uns nicht interessiert, was für einen Index der Unfall hatte, zu welchem genauen Datum ein Unfall stattgefunden hat, in welchem der zwei Jahre der Unfall war und auch die genauen geographischen Koordinaten sind nicht von Bedeutung.

```python
#Löschung von unbrauchbaren Spalten (axis = 1: Spalten löschen)
df = df.drop(["Accident_Index", "Accident Date", "Year", "Latitude", "Longitude"], axis = 1)
```

Wie bereits oben erwähnt, werden die Einträge von "Carriageway_Hazards" von Null zu "No Hazards" geändert, da dies aussagekräftiger ist, als Null in den Einträgen stehen zu lassen.

```python
#Ersetzung der Null Werte zu "No Hazards"
df["Carriageway_Hazards"] = df["Carriageway_Hazards"].fillna("No Hazards")
```

In den Spalten „Road_Surface_Conditions“, „Road_Type“, „Time“ und „Weather_Conditions“ fehlen einige Einträge. Dies lässt sich auch in der ursprünglichen Excel-Tabelle erkennen. Damit keine Datensätze verloren gehen, werden die leeren Felder nicht gelöscht, sondern durch den Eintrag „No Data“ ersetzt.

Eine Ersetzung durch Mittelwerte ist hier nicht sinnvoll, da es sich bei den meisten dieser Spalten um Merkmale handelt, für die sich kein sinnvoller Durchschnitt berechnen lässt. Außerdem könnten dadurch falsche Werte entstehen, die nicht der tatsächlichen Situation entsprechen. Beispielsweise wäre es nicht sinnvoll, eine fehlende Uhrzeit oder Geschwindigkeitsbegrenzung durch einen Durchschnittswert zu ersetzen, da an einer bestimmten Stelle bereits eine feste Geschwindigkeitsbegrenzung gilt. Durch die Verwendung von „No Data“ bleiben alle Datensätze erhalten und es ist trotzdem erkennbar, bei welchen Einträgen Informationen fehlen.

```python
#Ersetzung fehlender Werte durch "No Data"
df = df.fillna({"Road_Surface_Conditions": "No Data", "Road_Type": "No Data", "Time": "No Data", "Weather_Conditions": "No Data"})
```

Als letzten Schritt der Datenbereinigung werden die Uhrzeiten angepasst. Dabei werden die Minutenangaben entfernt, sodass nur noch die jeweilige Stunde erhalten bleibt. Dies ist für die weitere Analyse sinnvoller, da vor allem untersucht werden soll, zu welchen Uhrzeiten die meisten Unfälle passieren. Es ist dabei weniger wichtig, ob ein Unfall beispielsweise um 14:10 Uhr oder um 14:45 Uhr passiert ist. Beide Unfälle werden somit der gleichen Stunde zugeordnet. Dadurch lassen sich die Unfallzahlen besser nach Uhrzeiten vergleichen.

```python
#Funktion die eine gegebene Uhrzeit bei ":" teilt und die vorderste Zahl zurückgibt
def berechne_stunde(uhrzeit):
    #[0] am Ende wichtig, da ansonsten eine Liste im Format [XX, YY], bzw [X, YY] zurückgegeben wird
    return uhrzeit.split(":")[0]

#Ersetze alle alten Werte durch neue Werte, wobei auf jeden Eintrag die Funktion berechne_stunde angewandt wird
df["Time"] = df["Time"].apply(berechne_stunde)
```
