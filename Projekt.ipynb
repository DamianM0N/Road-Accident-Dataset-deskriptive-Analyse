Damit keine Datensätze verloren gehen, werden die leeren Felder nicht gelöscht, sondern durch den Eintrag „No Data“ ersetzt.

Eine Ersetzung durch Mittelwerte ist hier nicht sinnvoll, da es sich bei den meisten dieser Spalten um Merkmale handelt, für die sich kein sinnvoller Durchschnitt berechnen lässt. Außerdem könnten dadurch falsche Werte entstehen, die nicht der tatsächlichen Situation entsprechen. Beispielsweise wäre es nicht sinnvoll, eine fehlende Uhrzeit oder Geschwindigkeitsbegrenzung durch einen Durchschnittswert zu ersetzen, da an einer bestimmten Stelle bereits eine feste Geschwindigkeitsbegrenzung gilt.
```python
#Ersetzung fehlender Werte durch "No Data"
df = df.fillna({"Road_Surface_Conditions": "No Data", "Road_Type": "No Data", "Time": "No Data", "Weather_Conditions": "No Data"})
```

Für weitere Analyse erweist es sich als sinnvoll eine weitere Spalte zu der Tabelle hinzuzufügen. Bei dieser Spalte handelt es sich um die Spalte "Season" und deren Werte sind die vier Jahreszeiten. Dabei richtet sich die Einteilung der Monate an die meteorologische Jahreszeiten an. Außerdem werden die einzelnen Tage in "Weekday" und "Weekend" unterteilt, damit man im weiter Verlauf untersuchen kann, ob eine der Kategorien in Bezug auf Unfälle bevorzugt wird.

```python
#Zuweisung der Monate den meteorologischen Jahreszeiten
jahreszeiten_zuweisung = {
    "Dec": "Winter", "Jan": "Winter", "Feb": "Winter",
    "Mar": "Spring", "Apr": "Spring", "May": "Spring",
    "Jun": "Summer", "Jul": "Summer", "Aug": "Summer",
    "Sep": "Autumn", "Oct": "Autumn", "Nov": "Autumn"}

df["Season"] = df["Month"].map(jahreszeiten_zuweisung)

#Zuweisung der Tage
tage_zuweisung = {"Monday": "Weekday", "Tuesday": "Weekday", "Wednesday": "Weekday", "Thursday": "Weekday", "Friday": "Weekday", "Saturday": "Weekend", "Sunday": "Weekend"}

df["Week_Part"] = df["Day_of_Week"].map(tage_zuweisung)
```

Als letzten Schritt der Datenbereinigung werden die Uhrzeiten angepasst. Dabei werden die Minutenangaben entfernt, sodass nur noch die jeweilige Stunde erhalten bleibt. Dies ist für die weitere Analyse sinnvoller, da vor allem untersucht werden soll, zu welchen Uhrzeiten die meisten Unfälle passieren. Es ist dabei weniger wichtig, ob ein Unfall beispielsweise um 14:10 Uhr oder um 14:45 Uhr passiert ist. Beide Unfälle werden somit der gleichen Stunde zugeordnet. Dadurch lassen sich die Unfallzahlen besser nach Uhrzeiten vergleichen.  Ebenfalls wird noch nach Unterteilt, ob der Unfall Morgens, Mittags, Abends oder Nachts stattgefunden hat. Dazu wählen wir die folgende Einteilung:  Morgens 06:00 - 11:59 / Mittags 12.00 - 17:59 / Abends 18:00 - 21:59 / Nachts 22:00 - 05:59.

```python
#Funktion die eine gegebene Uhrzeit bei ":" teilt und die vorderste Zahl zurückgibt
def berechne_stunde(uhrzeit):
    #[0] am Ende wichtig, da ansonsten eine Liste im Format [XX, YY], bzw [X, YY] zurückgegeben wird
    return uhrzeit.split(":")[0]

#Ersetze alle alten Werte durch neue Werte, wobei auf jeden Eintrag die Funktion berechne_stunde angewandt wird
df["Time"] = df["Time"].apply(berechne_stunde)

#Funktion für die Zuweisung der Tageszeiten
def tageszeit(uhrzeit):
    if uhrzeit == "No Data":
        return "No Data"
    elif 6 <= int(uhrzeit) < 12:
        return "Morning"
    elif 12 <= int(uhrzeit) < 18:
        return "Afternoon"
    elif 18 <= int(uhrzeit) < 22:
        return "Evening"
    else:
        return "Night"

df["Day_Period"] = df["Time"].map(tageszeit)
```

```python
#Werte der Benford'schen Funktion
benford = pd.Series([30.1, 17.6, 12.5, 9.7, 7.9, 6.7, 5.8, 5.1, 4.6], index = [0, 1, 2, 3, 4, 5, 6, 7, 8])

fig, axes = plt.subplots(1, 3, figsize = (18, 6), dpi = 200)

#Berechnung der ersten Ziffern für die Anzahl der Verletzten
erste_ziffer = df["Number_of_Casualties"].astype(str).str[0]
#Angabe der Häufigkeit mit Rundung
haeufigkeit = round(erste_ziffer.value_counts(normalize = True).sort_index() * 100, 0)

ax0 = sns.barplot(ax = axes[0], x = haeufigkeit.index.astype(int), y = haeufigkeit.values)
sns.lineplot(ax = axes[0], x = benford.index, y = benford.values, marker = "o", color = "red", label = "Benford")
ax0.set_xlabel("")
ax0.set_ylabel("Anteil in %")
ax0.set_title("Vergleich mit Benford'schen Gesetz für Anzahl der Verletzten")

#Berechnung der ersten Ziffer für die Anzahl von beteiligten Fahrzeugen
erste_ziffer = df["Number_of_Vehicles"].astype(str).str[0]
#Angabe der Häufigkeit mit Rundung
haeufigkeit = round(erste_ziffer.value_counts(normalize = True).sort_index() * 100, 0)

ax1 = sns.barplot(ax = axes[1], x = haeufigkeit.index.astype(int), y = haeufigkeit.values)
sns.lineplot(ax = axes[1], x = benford.index, y = benford.values, marker = "o", color = "red", label = "Benford")
ax1.set_xlabel("")
ax1.set_ylabel("Anteil in %")
ax1.set_title("Vergleich mit Benford'schen Gesetz für Anzahl der Verletzten")

#Berechnung der ersten Ziffer für Geschwindigkeitsbegrenzungen
erste_ziffer = df["Speed_limit"].astype(str).str[0]
#Angabe der Häufigkeit mit Rundung
haeufigkeit = round(erste_ziffer.value_counts(normalize = True).sort_index() * 100, 0)

ax2 = sns.barplot(ax = axes[2], x = haeufigkeit.index.astype(int), y = haeufigkeit.values)
sns.lineplot(ax = axes[2], x = benford.index, y = benford.values, marker = "o", color = "red", label = "Benford")
ax2.set_xlabel("")
ax2.set_ylabel("Anteil in %")
ax2.set_title("Vergleich mit Bernford'schen Gesetz für die Geschwindigkeitsbegrenzungen")

plt.tight_layout()
plt.show()
```

Zu guter Letzt werden noch die andern Variablen anhand ihrer Korrelationen untersucht.

```python
#Funktion welche die verschiedenen Cramer V's berechnet
def cramers_v_matrix(df, cols):
    #Erstellung einer Matrix die mit 1-en gefüll wird
    matrix = pd.DataFrame(np.ones((len(cols), len(cols))), index = cols, columns = cols)
    #Iterationen über die Spalten
    for col1 in cols:
        for col2 in cols:
            if col1 != col2:
                ct = pd.crosstab(df[col1], df[col2])
                matrix.loc[col1, col2] = association(ct, method = "cramer")
    return matrix

nom_cols = ['Road_Type', 'Weather_Conditions', 'Road_Surface_Conditions', 'Urban_or_Rural_Area', 'Month', 'Day_of_Week', 'Time', 'Carriageway_Hazards', 'Junction_Control', 'Light_Conditions', 'Vehicle_Type', 'Junction_Detail'] 
cramer_df = cramers_v_matrix(df, nom_cols)
cramer_df
```

Da es sich bei der Unfallschwere aufgrund ihrer natürlichen Rangfolge um eine ordinalskalierte Variable handelt, wurde sie bei der Berechnung von Cramér’s V ausgeklammert und die Analyse gezielt auf die zwölf nominalskalierten Merkmale beschränkt. Die resultierende Matrix zeigt, dass zwischen den meisten dieser Einflussfaktoren nur schwache Zusammenhänge bestehen, während vor allem die Variablen moderat bis deutlich miteinander verknüpft sind, die auch inhaltlich eng zusammenhängen. Die stärksten Zusammenhänge finden sich im Bereich der Straßen und Kreuzungsinfrastruktur, insbesondere zwischen der Verkehrsregelung (Junction_Control) und der Art des Knotenpunkts (Junction_Detail) mit einem Wert von rund 0.49 sowie zwischen dem Straßentyp (Road_Type) und dem Knotenpunkt (Junction_Detail) mit etwa 0.40. Ein weiterer deutlicher Zusammenhang besteht zwischen den Wetterbedingungen und dem Fahrbahnzustand (Road_Surface_Conditions) mit rund 0.44, gefolgt von der Verknüpfung zwischen der Uhrzeit  und den Lichtverhältnissen mit etwa 0.37. Zudem zeigen sich beim Vergleich zwischen städtischem und ländlichem Raum (Urban_or_Rural_Area) moderate Zusammenhänge mit den Lichtverhältnissen ca. 0.33, den Knotenpunkten ca. 0,32 und der Verkehrsregelung ca. 0,31, während saisonale Einflüsse über den Monat vor allem beim Fahrbahnzustand ca. 0.24, den Lichtverhältnissen ca. 0.16 und dem Wetter ca. 0.13 sichtbar werden. Merkmale wie der Fahrzeugtyp (Vehicle_Type, durchgehend unter 0.02) oder der Wochentag (überwiegend unter 0.05) weisen hingegen kaum statistische Zusammenhänge zu den übrigen Kategorien auf.   

Logisch gesehen ergeben diese Muster ein sehr stimmiges Bild, da die untersuchten Variablen in der realen Welt voneinander abhängen oder gemeinsam durch übergeordnete Rahmenbedingungen wie Ort, Zeit oder Wetter geprägt werden. Im Bereich der Straßen und Kreuzungsinfrastruktur greifen Knotenpunktart, Verkehrsregelung und Straßentyp ineinander, weil eine Straße baulich und rechtlich auf die jeweilige Verkehrssituation zugeschnitten wird, ein Kreisverkehr, eine Hauptverkehrsstraße oder eine Einmündung bringt jeweils typische Verkehrsregelungen und Straßentypen mit sich. Ähnlich verhält es sich beim Wetter und dem Fahrbahnzustand: Regen, Schnee oder Frost machen die Straße nass, glatt oder vereist, wobei der Zusammenhang nicht perfekt ist, weil eine Fahrbahn nach einem Schauer noch nass sein kann, obwohl es bereits nicht mehr regnet, oder umgekehrt leichter Niederschlag auf warmem Asphalt kaum Wirkung zeigt. Auch die Uhrzeit und die Lichtverhältnisse sind logisch eng verknüpft, weil es morgens hell und abends sowie nachts dunkel wird, dass der Wert dennoch nicht bei 1.0 liegt, erklärt sich unter anderem durch jahreszeitliche Unterschiede bei den Sonnenauf und untergangszeiten sowie durch Straßenbeleuchtung. Beim Vergleich zwischen städtischem und ländlichem Raum spiegeln sich wiederum die unterschiedlichen baulichen Gegebenheiten wider, da innerorts häufiger Kreuzungen, Ampeln und Straßenbeleuchtung vorhanden sind, während außerorts Straßen oft offener verlaufen und anders geregelt sind. Der Monat steht stellvertretend für die Jahreszeit und beeinflusst deshalb genau dort die Werte, wo saisonale Effekte wie kürzere Tage im Winter oder winterliche Straßenverhältnisse auftreten. Umgekehrt ist es ebenso logisch, dass Merkmale wie Fahrzeugtyp oder Wochentag kaum mit den übrigen Kategorien zusammenhängen: Verschiedene Fahrzeugarten sind grundsätzlich jeden Tag und auf vielen Strecken unterwegs, und der Wochentag verändert zwar das allgemeine Verkehrsaufkommen und den Wegezweck, nicht aber die bauliche Gestaltung einer Kreuzung oder die physikalische Wirkung von Nässe auf der Fahrbahn. Zusammengefasst entstehen stärkere Zusammenhänge in der Matrix immer dann, wenn ein Merkmal die unmittelbare Voraussetzung oder den baulichen beziehungsweise zeitlichen Rahmen für ein anderes bildet.<br><br>

Um den Zusammenhang zwischen der ordinalskalierten Unfallschwere und den kardinalen Merkmalen zu untersuchen, wird die Unfallschwere zunächst rangbasiert codiert (0 = Slight, 1 = Serious, 2 = Fatal) und anschließend die Spearman-Rangkorrelation berechnet.

```python
#Codierung der Severity zu Zahlen
severity_map = {"Slight": 0, "Serious": 1, "Fatal": 2}
#Einfügen einer neuen Spalte
df["Severity_Code"] = df["Accident_Severity"].map(severity_map)

#Berechnunug der Matrix
corr_spearman = df[["Severity_Code", "Number_of_Casualties", "Number_of_Vehicles", "Speed_limit"]].corr(method = "spearman")

corr_spearman
```

Die Ergebnisse zeigen, dass die Zusammenhänge zwischen Severity_Code und den betrachteten numerischen Variablen insgesamt nur schwach ausgeprägt sind. Mit der Geschwindigkeitsbegrenzung ca. 0.07 sowie der Anzahl der Verletzten ca. 0.05 ergeben sich lediglich sehr geringe positive Korrelationen. Zur Anzahl der beteiligten Fahrzeuge besteht mit rund -0.11 sogar ein leicht negativer Zusammenhang, was sich mit den vorherigen Beobachtungen deckt, wonach Unfälle mit vielen beteiligten Fahrzeugen häufiger glimpflicher verlaufen als einzelne Unfälle mit wenigen Beteiligten. Insgesamt verdeutlicht die Auswertung, dass sich die Unfallschwere nicht allein durch einzelne kardinale Kennzahlen wie die Fahrzeug  oder Verletztenzahl erklären lässt.<br><br>
